```python
# ============================================================
# STEP 9: DATE AND TIME FEATURE ENGINEERING
# Smart City Traffic & Accident Analytics
# ============================================================

# ---------- TRAFFIC DATA ----------

traffic_df['Date'] = traffic_df['Date_Time'].dt.date
traffic_df['Year'] = traffic_df['Date_Time'].dt.year
traffic_df['Month'] = traffic_df['Date_Time'].dt.month
traffic_df['Month_Name'] = traffic_df['Date_Time'].dt.month_name()
traffic_df['Day'] = traffic_df['Date_Time'].dt.day
traffic_df['Day_Name'] = traffic_df['Date_Time'].dt.day_name()
traffic_df['Week'] = traffic_df['Date_Time'].dt.isocalendar().week.astype(int)
traffic_df['Quarter'] = traffic_df['Date_Time'].dt.quarter
traffic_df['Hour'] = traffic_df['Date_Time'].dt.hour

traffic_df['Is_Weekend'] = traffic_df['Date_Time'].dt.dayofweek >= 5


# ---------- ACCIDENT DATA ----------

accident_df['Date'] = accident_df['Date_Time'].dt.date
accident_df['Year'] = accident_df['Date_Time'].dt.year
accident_df['Month'] = accident_df['Date_Time'].dt.month
accident_df['Month_Name'] = accident_df['Date_Time'].dt.month_name()
accident_df['Day'] = accident_df['Date_Time'].dt.day
accident_df['Day_Name'] = accident_df['Date_Time'].dt.day_name()
accident_df['Week'] = accident_df['Date_Time'].dt.isocalendar().week.astype(int)
accident_df['Quarter'] = accident_df['Date_Time'].dt.quarter
accident_df['Hour'] = accident_df['Date_Time'].dt.hour

accident_df['Is_Weekend'] = accident_df['Date_Time'].dt.dayofweek >= 5


print("Traffic Dataset:")
display(
    traffic_df[
        ['Date_Time', 'Date', 'Year', 'Month', 'Month_Name',
         'Day', 'Day_Name', 'Week', 'Quarter', 'Hour', 'Is_Weekend']
    ].head()
)

print("\nAccident Dataset:")
display(
    accident_df[
        ['Date_Time', 'Date', 'Year', 'Month', 'Month_Name',
         'Day', 'Day_Name', 'Week', 'Quarter', 'Hour', 'Is_Weekend']
    ].head()
)
```

```python
# ============================================================
# STEP 10: TIME PERIOD AND PEAK HOUR FEATURE ENGINEERING
# Smart City Traffic & Accident Analytics
# ============================================================

def get_time_period(hour):
    if 0 <= hour <= 5:
        return 'Night'
    elif 6 <= hour <= 11:
        return 'Morning'
    elif 12 <= hour <= 16:
        return 'Afternoon'
    elif 17 <= hour <= 20:
        return 'Evening'
    else:
        return 'Late Evening'


def get_peak_hour(hour):
    if 7 <= hour <= 9 or 17 <= hour <= 19:
        return 'Peak'
    else:
        return 'Non-Peak'


# Apply to Traffic Dataset
traffic_df['Time_Period'] = traffic_df['Hour'].apply(get_time_period)
traffic_df['Peak_Hour'] = traffic_df['Hour'].apply(get_peak_hour)


# Apply to Accident Dataset
accident_df['Time_Period'] = accident_df['Hour'].apply(get_time_period)
accident_df['Peak_Hour'] = accident_df['Hour'].apply(get_peak_hour)


print("========== TRAFFIC DATA ==========")

print("\nTime Period Distribution:")
print(traffic_df['Time_Period'].value_counts())

print("\nPeak Hour Distribution:")
print(traffic_df['Peak_Hour'].value_counts())


print("\n\n========== ACCIDENT DATA ==========")

print("\nTime Period Distribution:")
print(accident_df['Time_Period'].value_counts())

print("\nPeak Hour Distribution:")
print(accident_df['Peak_Hour'].value_counts())


print("\n\nSample Traffic Records:")
display(
    traffic_df[
        ['Date_Time', 'Hour', 'Time_Period', 'Peak_Hour']
    ].head(15)
)
```

```python
# ============================================================
# STEP 11: CONGESTION INDEX FEATURE ENGINEERING
# Smart City Traffic & Accident Analytics
# ============================================================

# Calculate Congestion Index
traffic_df['Congestion_Index'] = (
    traffic_df['Vehicle_Count'] / traffic_df['Average_Speed']
)

# Create categories based on the Congestion Index distribution
traffic_df['Congestion_Category'] = pd.qcut(
    traffic_df['Congestion_Index'],
    q=3,
    labels=['Low', 'Moderate', 'High']
)


print("========== CONGESTION INDEX ==========")

print("\nCongestion Index Statistics:")
print(traffic_df['Congestion_Index'].describe())

print("\nCongestion Category Distribution:")
print(traffic_df['Congestion_Category'].value_counts().sort_index())


print("\nSample Records:")
display(
    traffic_df[
        [
            'Location',
            'Date_Time',
            'Vehicle_Count',
            'Average_Speed',
            'Congestion_Level',
            'Congestion_Index',
            'Congestion_Category'
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 12: ACCIDENT SEVERITY SCORE FEATURE ENGINEERING
# Smart City Traffic & Accident Analytics
# ============================================================

# Map accident severity to an analytical numeric score
severity_score_map = {
    'Minor': 1,
    'Moderate': 2,
    'Severe': 3,
    'Fatal': 4
}

accident_df['Accident_Severity_Score'] = (
    accident_df['Accident_Severity'].map(severity_score_map)
)

print("========== ACCIDENT SEVERITY SCORE ==========")

print("\nSeverity Score Distribution:")
print(
    accident_df[
        ['Accident_Severity', 'Accident_Severity_Score']
    ].drop_duplicates().sort_values('Accident_Severity_Score')
)

print("\nAccident Severity Counts:")
print(
    accident_df['Accident_Severity']
    .value_counts()
)

print("\nAccident Severity Score Statistics:")
print(
    accident_df['Accident_Severity_Score'].describe()
)

print("\nSample Records:")
display(
    accident_df[
        [
            'Accident_ID',
            'Date_Time',
            'Location',
            'Accident_Severity',
            'Accident_Severity_Score',
            'Number_of_Vehicles',
            'Casualties',
            'Traffic_Density'
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 13: CREATE DATE DIMENSION
# Smart City Traffic & Accident Analytics
# ============================================================

# Combine dates from both datasets
all_dates = pd.concat([
    traffic_df['Date_Time'].dt.normalize(),
    accident_df['Date_Time'].dt.normalize()
]).drop_duplicates().sort_values()

# Create Date Dimension
dim_date = pd.DataFrame({
    'Full_Date': all_dates
})

dim_date['Date_Key'] = (
    dim_date['Full_Date'].dt.strftime('%Y%m%d').astype(int)
)

dim_date['Day'] = dim_date['Full_Date'].dt.day
dim_date['Day_Name'] = dim_date['Full_Date'].dt.day_name()
dim_date['Week'] = dim_date['Full_Date'].dt.isocalendar().week.astype(int)
dim_date['Month'] = dim_date['Full_Date'].dt.month
dim_date['Month_Name'] = dim_date['Full_Date'].dt.month_name()
dim_date['Quarter'] = dim_date['Full_Date'].dt.quarter
dim_date['Year'] = dim_date['Full_Date'].dt.year
dim_date['Is_Weekend'] = (
    dim_date['Full_Date'].dt.dayofweek >= 5
)

# Reorder columns
dim_date = dim_date[
    [
        'Date_Key',
        'Full_Date',
        'Day',
        'Day_Name',
        'Week',
        'Month',
        'Month_Name',
        'Quarter',
        'Year',
        'Is_Weekend'
    ]
]

print("========== DATE DIMENSION ==========")

print("\nShape:")
print(dim_date.shape)

print("\nDate Range:")
print(dim_date['Full_Date'].min(), "to", dim_date['Full_Date'].max())

print("\nColumns:")
print(dim_date.columns.tolist())

print("\nSample Records:")
display(dim_date.head(10))

print("\nDate Dimension Validation:")
print("Duplicate Date_Key:", dim_date['Date_Key'].duplicated().sum())
print("Missing Values:")
print(dim_date.isnull().sum())
```

```python
# ============================================================
# STEP 14: CREATE TIME DIMENSION
# Smart City Traffic & Accident Analytics
# ============================================================

# Create one record for each hour of the day
dim_time = pd.DataFrame({
    'Hour': range(24)
})

# Create time dimension attributes
dim_time['Time_Key'] = dim_time['Hour']
dim_time['Time'] = pd.to_datetime(
    dim_time['Hour'].astype(str).str.zfill(2) + ':00',
    format='%H:%M'
).dt.strftime('%H:%M:%S')

dim_time['Hour_Label'] = (
    dim_time['Hour'].astype(str).str.zfill(2) + ':00'
)

# Reuse the time-period and peak-hour definitions
dim_time['Time_Period'] = dim_time['Hour'].apply(
    get_time_period
)

dim_time['Peak_Hour'] = dim_time['Hour'].apply(
    get_peak_hour
)

# Arrange columns
dim_time = dim_time[
    [
        'Time_Key',
        'Time',
        'Hour',
        'Hour_Label',
        'Time_Period',
        'Peak_Hour'
    ]
]

print("========== TIME DIMENSION ==========")

print("\nShape:")
print(dim_time.shape)

print("\nColumns:")
print(dim_time.columns.tolist())

print("\nTime Period Distribution:")
print(dim_time['Time_Period'].value_counts())

print("\nPeak Hour Distribution:")
print(dim_time['Peak_Hour'].value_counts())

print("\nSample Records:")
display(dim_time.head(10))

print("\nLast 5 Records:")
display(dim_time.tail(5))

print("\nTime Dimension Validation:")
print("Duplicate Time_Key:", dim_time['Time_Key'].duplicated().sum())
print("Missing Values:")
print(dim_time.isnull().sum())
```

```python
# ============================================================
# STEP 15: CREATE WEATHER DIMENSION
# Smart City Traffic & Accident Analytics
# ============================================================

# Extract unique weather conditions
dim_weather = pd.DataFrame({
    'Weather_Condition': sorted(
        accident_df['Weather_Condition'].unique()
    )
})

# Create surrogate key
dim_weather.insert(
    0,
    'Weather_Key',
    range(1, len(dim_weather) + 1)
)

print("========== WEATHER DIMENSION ==========")

print("\nShape:")
print(dim_weather.shape)

print("\nColumns:")
print(dim_weather.columns.tolist())

print("\nWeather Conditions:")
print(dim_weather['Weather_Condition'].tolist())

print("\nSample Records:")
display(dim_weather)

print("\nWeather Dimension Validation:")
print(
    "Duplicate Weather_Key:",
    dim_weather['Weather_Key'].duplicated().sum()
)

print(
    "Duplicate Weather_Condition:",
    dim_weather['Weather_Condition'].duplicated().sum()
)

print("\nMissing Values:")
print(dim_weather.isnull().sum())
```

```python
# ============================================================
# STEP 16: CREATE ROAD CONDITION DIMENSION
# Smart City Traffic & Accident Analytics
# ============================================================

# Extract unique road conditions
dim_road_condition = pd.DataFrame({
    'Road_Condition': sorted(
        accident_df['Road_Condition'].unique()
    )
})

# Create surrogate key
dim_road_condition.insert(
    0,
    'Road_Condition_Key',
    range(1, len(dim_road_condition) + 1)
)

print("========== ROAD CONDITION DIMENSION ==========")

print("\nShape:")
print(dim_road_condition.shape)

print("\nColumns:")
print(dim_road_condition.columns.tolist())

print("\nRoad Conditions:")
print(dim_road_condition['Road_Condition'].tolist())

print("\nSample Records:")
display(dim_road_condition)

print("\nRoad Condition Dimension Validation:")
print(
    "Duplicate Road_Condition_Key:",
    dim_road_condition['Road_Condition_Key'].duplicated().sum()
)

print(
    "Duplicate Road_Condition:",
    dim_road_condition['Road_Condition'].duplicated().sum()
)

print("\nMissing Values:")
print(dim_road_condition.isnull().sum())
```

```python
# ============================================================
# STEP 17: CREATE VEHICLE TYPE DIMENSION
# Smart City Traffic & Accident Analytics
# ============================================================

# Extract unique vehicle types
dim_vehicle_type = pd.DataFrame({
    'Vehicle_Type': sorted(
        accident_df['Vehicle_Type'].unique()
    )
})

# Create surrogate key
dim_vehicle_type.insert(
    0,
    'Vehicle_Type_Key',
    range(1, len(dim_vehicle_type) + 1)
)

print("========== VEHICLE TYPE DIMENSION ==========")

print("\nShape:")
print(dim_vehicle_type.shape)

print("\nColumns:")
print(dim_vehicle_type.columns.tolist())

print("\nVehicle Types:")
print(dim_vehicle_type['Vehicle_Type'].tolist())

print("\nSample Records:")
display(dim_vehicle_type)

print("\nVehicle Type Dimension Validation:")
print(
    "Duplicate Vehicle_Type_Key:",
    dim_vehicle_type['Vehicle_Type_Key'].duplicated().sum()
)

print(
    "Duplicate Vehicle_Type:",
    dim_vehicle_type['Vehicle_Type'].duplicated().sum()
)

print("\nMissing Values:")
print(dim_vehicle_type.isnull().sum())
```

```python
# ============================================================
# STEP 18: CREATE FACT TRAFFIC TABLE
# Smart City Traffic & Accident Analytics
# ============================================================

# Create Fact_Traffic from traffic sensor data
fact_traffic = traffic_df[
    [
        'Date_Time',
        'Sensor_ID',
        'Location',
        'Vehicle_Count',
        'Average_Speed',
        'Congestion_Level',
        'Congestion_Index',
        'Congestion_Category'
    ]
].copy()

# Create Date_Key
fact_traffic['Date_Key'] = (
    fact_traffic['Date_Time']
    .dt.strftime('%Y%m%d')
    .astype(int)
)

# Create Time_Key
fact_traffic['Time_Key'] = (
    fact_traffic['Date_Time'].dt.hour
)

# Create a surrogate Fact key
fact_traffic.insert(
    0,
    'Traffic_Key',
    range(1, len(fact_traffic) + 1)
)

# Arrange columns according to the dimensional model
fact_traffic = fact_traffic[
    [
        'Traffic_Key',
        'Date_Key',
        'Time_Key',
        'Sensor_ID',
        'Location',
        'Vehicle_Count',
        'Average_Speed',
        'Congestion_Level',
        'Congestion_Index',
        'Congestion_Category',
        'Date_Time'
    ]
]

print("========== FACT TRAFFIC ==========")

print("\nShape:")
print(fact_traffic.shape)

print("\nColumns:")
print(fact_traffic.columns.tolist())

print("\nSample Records:")
display(fact_traffic.head(10))

print("\nFact Traffic Validation:")
print(
    "Duplicate Traffic_Key:",
    fact_traffic['Traffic_Key'].duplicated().sum()
)

print(
    "Missing Values:"
)
print(fact_traffic.isnull().sum())

print("\nDate Key Validation:")
print(
    "Invalid Date_Key:",
    (~fact_traffic['Date_Key'].isin(dim_date['Date_Key'])).sum()
)

print("\nTime Key Validation:")
print(
    "Invalid Time_Key:",
    (~fact_traffic['Time_Key'].isin(dim_time['Time_Key'])).sum()
)
```

```python
# ============================================================
# STEP 19: CREATE FACT ACCIDENT TABLE
# Smart City Traffic & Accident Analytics
# ============================================================

# Start with the required accident fields
fact_accident = accident_df[
    [
        'Date_Time',
        'Accident_ID',
        'Location',
        'Weather_Condition',
        'Road_Condition',
        'Vehicle_Type',
        'Accident_Severity',
        'Accident_Severity_Score',
        'Number_of_Vehicles',
        'Casualties',
        'Traffic_Density'
    ]
].copy()

# ------------------------------------------------------------
# Create Date_Key
# ------------------------------------------------------------
fact_accident['Date_Key'] = (
    fact_accident['Date_Time']
    .dt.strftime('%Y%m%d')
    .astype(int)
)

# ------------------------------------------------------------
# Create Time_Key
# ------------------------------------------------------------
fact_accident['Time_Key'] = (
    fact_accident['Date_Time'].dt.hour
)

# ------------------------------------------------------------
# Map Weather to Weather_Key
# ------------------------------------------------------------
weather_map = dict(
    zip(
        dim_weather['Weather_Condition'],
        dim_weather['Weather_Key']
    )
)

fact_accident['Weather_Key'] = (
    fact_accident['Weather_Condition'].map(weather_map)
)

# ------------------------------------------------------------
# Map Road Condition to Road_Condition_Key
# ------------------------------------------------------------
road_condition_map = dict(
    zip(
        dim_road_condition['Road_Condition'],
        dim_road_condition['Road_Condition_Key']
    )
)

fact_accident['Road_Condition_Key'] = (
    fact_accident['Road_Condition'].map(
        road_condition_map
    )
)

# ------------------------------------------------------------
# Map Vehicle Type to Vehicle_Type_Key
# ------------------------------------------------------------
vehicle_type_map = dict(
    zip(
        dim_vehicle_type['Vehicle_Type'],
        dim_vehicle_type['Vehicle_Type_Key']
    )
)

fact_accident['Vehicle_Type_Key'] = (
    fact_accident['Vehicle_Type'].map(
        vehicle_type_map
    )
)

# ------------------------------------------------------------
# Create surrogate Fact key
# ------------------------------------------------------------
fact_accident.insert(
    0,
    'Accident_Key',
    range(1, len(fact_accident) + 1)
)

# ------------------------------------------------------------
# Arrange columns according to dimensional model
# ------------------------------------------------------------
fact_accident = fact_accident[
    [
        'Accident_Key',
        'Date_Key',
        'Time_Key',
        'Accident_ID',
        'Location',
        'Weather_Key',
        'Road_Condition_Key',
        'Vehicle_Type_Key',
        'Accident_Severity',
        'Accident_Severity_Score',
        'Number_of_Vehicles',
        'Casualties',
        'Traffic_Density',
        'Date_Time'
    ]
]

print("========== FACT ACCIDENT ==========")

print("\nShape:")
print(fact_accident.shape)

print("\nColumns:")
print(fact_accident.columns.tolist())

print("\nSample Records:")
display(fact_accident.head(10))

print("\nFact Accident Validation:")

print(
    "Duplicate Accident_Key:",
    fact_accident['Accident_Key'].duplicated().sum()
)

print("\nMissing Values:")
print(fact_accident.isnull().sum())

print("\nDate Key Validation:")
print(
    "Invalid Date_Key:",
    (~fact_accident['Date_Key'].isin(
        dim_date['Date_Key']
    )).sum()
)

print("\nTime Key Validation:")
print(
    "Invalid Time_Key:",
    (~fact_accident['Time_Key'].isin(
        dim_time['Time_Key']
    )).sum()
)

print("\nWeather Key Validation:")
print(
    "Invalid Weather_Key:",
    (~fact_accident['Weather_Key'].isin(
        dim_weather['Weather_Key']
    )).sum()
)

print("\nRoad Condition Key Validation:")
print(
    "Invalid Road_Condition_Key:",
    (~fact_accident['Road_Condition_Key'].isin(
        dim_road_condition['Road_Condition_Key']
    )).sum()
)

print("\nVehicle Type Key Validation:")
print(
    "Invalid Vehicle_Type_Key:",
    (~fact_accident['Vehicle_Type_Key'].isin(
        dim_vehicle_type['Vehicle_Type_Key']
    )).sum()
)
```

```python
# ============================================================
# STEP 20: FINAL DIMENSIONAL MODEL VALIDATION
# Smart City Traffic & Accident Analytics
# ============================================================

print("=" * 65)
print("FINAL DIMENSIONAL MODEL VALIDATION")
print("=" * 65)

# ------------------------------------------------------------
# 1. Table Summary
# ------------------------------------------------------------
table_summary = pd.DataFrame({
    'Table': [
        'Dim_Date',
        'Dim_Time',
        'Dim_Weather',
        'Dim_Road_Condition',
        'Dim_Vehicle_Type',
        'Fact_Traffic',
        'Fact_Accident'
    ],
    'Rows': [
        len(dim_date),
        len(dim_time),
        len(dim_weather),
        len(dim_road_condition),
        len(dim_vehicle_type),
        len(fact_traffic),
        len(fact_accident)
    ],
    'Columns': [
        len(dim_date.columns),
        len(dim_time.columns),
        len(dim_weather.columns),
        len(dim_road_condition.columns),
        len(dim_vehicle_type.columns),
        len(fact_traffic.columns),
        len(fact_accident.columns)
    ]
})

print("\n========== TABLE SUMMARY ==========")
display(table_summary)

# ------------------------------------------------------------
# 2. Primary Key Validation
# ------------------------------------------------------------
print("\n========== PRIMARY KEY VALIDATION ==========")

pk_checks = {
    'Dim_Date.Date_Key': dim_date['Date_Key'].duplicated().sum(),
    'Dim_Time.Time_Key': dim_time['Time_Key'].duplicated().sum(),
    'Dim_Weather.Weather_Key': dim_weather['Weather_Key'].duplicated().sum(),
    'Dim_Road_Condition.Road_Condition_Key':
        dim_road_condition['Road_Condition_Key'].duplicated().sum(),
    'Dim_Vehicle_Type.Vehicle_Type_Key':
        dim_vehicle_type['Vehicle_Type_Key'].duplicated().sum(),
    'Fact_Traffic.Traffic_Key':
        fact_traffic['Traffic_Key'].duplicated().sum(),
    'Fact_Accident.Accident_Key':
        fact_accident['Accident_Key'].duplicated().sum()
}

for key, duplicates in pk_checks.items():
    status = "VALID" if duplicates == 0 else "INVALID"
    print(f"{key:<45} {status} | Duplicates: {duplicates}")

# ------------------------------------------------------------
# 3. Foreign Key Validation
# ------------------------------------------------------------
print("\n========== FOREIGN KEY VALIDATION ==========")

fk_checks = {
    'Fact_Traffic.Date_Key → Dim_Date':
        (~fact_traffic['Date_Key'].isin(dim_date['Date_Key'])).sum(),

    'Fact_Traffic.Time_Key → Dim_Time':
        (~fact_traffic['Time_Key'].isin(dim_time['Time_Key'])).sum(),

    'Fact_Accident.Date_Key → Dim_Date':
        (~fact_accident['Date_Key'].isin(dim_date['Date_Key'])).sum(),

    'Fact_Accident.Time_Key → Dim_Time':
        (~fact_accident['Time_Key'].isin(dim_time['Time_Key'])).sum(),

    'Fact_Accident.Weather_Key → Dim_Weather':
        (~fact_accident['Weather_Key'].isin(
            dim_weather['Weather_Key']
        )).sum(),

    'Fact_Accident.Road_Condition_Key → Dim_Road_Condition':
        (~fact_accident['Road_Condition_Key'].isin(
            dim_road_condition['Road_Condition_Key']
        )).sum(),

    'Fact_Accident.Vehicle_Type_Key → Dim_Vehicle_Type':
        (~fact_accident['Vehicle_Type_Key'].isin(
            dim_vehicle_type['Vehicle_Type_Key']
        )).sum()
}

for relationship, invalid in fk_checks.items():
    status = "VALID" if invalid == 0 else "INVALID"
    print(f"{relationship:<55} {status} | Invalid: {invalid}")

# ------------------------------------------------------------
# 4. Missing Value Validation
# ------------------------------------------------------------
print("\n========== MISSING VALUE VALIDATION ==========")

tables = {
    'Dim_Date': dim_date,
    'Dim_Time': dim_time,
    'Dim_Weather': dim_weather,
    'Dim_Road_Condition': dim_road_condition,
    'Dim_Vehicle_Type': dim_vehicle_type,
    'Fact_Traffic': fact_traffic,
    'Fact_Accident': fact_accident
}

for table_name, df in tables.items():
    missing = df.isnull().sum().sum()
    status = "VALID" if missing == 0 else "CHECK"
    print(f"{table_name:<25} {status} | Total Missing Values: {missing}")

# ------------------------------------------------------------
# 5. Overall Model Status
# ------------------------------------------------------------
all_pk_valid = all(value == 0 for value in pk_checks.values())
all_fk_valid = all(value == 0 for value in fk_checks.values())
all_missing_valid = all(
    df.isnull().sum().sum() == 0
    for df in tables.values()
)

print("\n" + "=" * 65)

if all_pk_valid and all_fk_valid and all_missing_valid:
    print("OVERALL MODEL STATUS: VALID")
    print("All dimensions and fact tables are ready for MySQL loading.")
else:
    print("OVERALL MODEL STATUS: CHECK REQUIRED")

print("=" * 65)
```

```python
# ============================================================
# STEP 89: PREPARE PREDICTIVE MODELING DATASET
# ============================================================

# Create a copy of the accident dataset
model_df = accident_df.copy()

# Select predictive features and target
model_df = model_df[
    [
        'Location',
        'Weather_Condition',
        'Road_Condition',
        'Vehicle_Type',
        'Traffic_Density',
        'Hour',
        'Time_Period',
        'Peak_Hour',
        'Month',
        'Day_Name',
        'Is_Weekend',
        'Number_of_Vehicles',
        'Casualties',
        'Accident_Severity',
        'Accident_Severity_Score'
    ]
].copy()

print("=" * 70)
print("PREDICTIVE MODELING DATASET")
print("=" * 70)

print("\nDataset Shape:")
print(model_df.shape)

print("\nColumns:")
print(model_df.columns.tolist())

print("\nMissing Values:")
print(model_df.isnull().sum())

print("\nAccident Severity Distribution:")
print(model_df['Accident_Severity'].value_counts())

print("\nAccident Severity Percentage:")
print(
    (model_df['Accident_Severity'].value_counts(normalize=True) * 100)
    .round(2)
)

print("\nSeverity Score Distribution:")
print(model_df['Accident_Severity_Score'].value_counts().sort_index())

print("\nSample Records:")
display(model_df.head())
```

```python
# ============================================================
# STEP 90: CREATE HIGH-RISK ACCIDENT TARGET
# ============================================================

# Create a modeling copy
risk_df = model_df.copy()

# Create binary target:
# 1 = High Risk (Severe or Fatal)
# 0 = Lower Risk (Minor or Moderate)

risk_df['High_Risk'] = risk_df['Accident_Severity'].isin(
    ['Severe', 'Fatal']
).astype(int)

# Check target distribution
print("=" * 70)
print("HIGH-RISK ACCIDENT TARGET")
print("=" * 70)

print("\nTarget Definition:")
print("1 = Severe or Fatal")
print("0 = Minor or Moderate")

print("\nTarget Distribution:")
print(
    risk_df['High_Risk']
    .value_counts()
    .sort_index()
)

print("\nTarget Percentage:")
print(
    (risk_df['High_Risk']
     .value_counts(normalize=True)
     .sort_index() * 100)
    .round(2)
)

print("\nTarget by Accident Severity:")
print(
    pd.crosstab(
        risk_df['Accident_Severity'],
        risk_df['High_Risk']
    )
)

print("\nModeling Dataset Shape:")
print(risk_df.shape)

print("\nSample:")
display(
    risk_df[
        [
            'Location',
            'Weather_Condition',
            'Road_Condition',
            'Vehicle_Type',
            'Traffic_Density',
            'Hour',
            'Time_Period',
            'Peak_Hour',
            'Month',
            'Day_Name',
            'Is_Weekend',
            'Number_of_Vehicles',
            'Accident_Severity',
            'High_Risk'
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 91: TRAIN-TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

# Predictive features
feature_columns = [
    'Location',
    'Weather_Condition',
    'Road_Condition',
    'Vehicle_Type',
    'Traffic_Density',
    'Hour',
    'Time_Period',
    'Peak_Hour',
    'Month',
    'Day_Name',
    'Is_Weekend',
    'Number_of_Vehicles'
]

X = risk_df[feature_columns].copy()
y = risk_df['High_Risk'].copy()

# Stratified 80/20 train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("=" * 70)
print("TRAIN-TEST SPLIT")
print("=" * 70)

print("\nOriginal Dataset:")
print("X:", X.shape)
print("y:", y.shape)

print("\nTraining Dataset:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting Dataset:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts().sort_index())

print("\nTraining Target Percentage:")
print(
    (y_train.value_counts(normalize=True).sort_index() * 100)
    .round(2)
)

print("\nTesting Target Distribution:")
print(y_test.value_counts().sort_index())

print("\nTesting Target Percentage:")
print(
    (y_test.value_counts(normalize=True).sort_index() * 100)
    .round(2)
)
```

```python
# ============================================================
# STEP 92: ENCODE CATEGORICAL VARIABLES
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Categorical features
categorical_features = [
    'Location',
    'Weather_Condition',
    'Road_Condition',
    'Vehicle_Type',
    'Traffic_Density',
    'Time_Period',
    'Peak_Hour',
    'Day_Name'
]

# Numerical features
numerical_features = [
    'Hour',
    'Month',
    'Is_Weekend',
    'Number_of_Vehicles'
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_features
        ),
        (
            'numerical',
            'passthrough',
            numerical_features
        )
    ]
)

# Fit preprocessing only on training data
X_train_encoded = preprocessor.fit_transform(X_train)

# Transform test data using the same fitted preprocessing
X_test_encoded = preprocessor.transform(X_test)

# Get encoded feature names
encoded_feature_names = preprocessor.get_feature_names_out()

print("=" * 70)
print("CATEGORICAL ENCODING")
print("=" * 70)

print("\nOriginal Training Shape:")
print(X_train.shape)

print("\nEncoded Training Shape:")
print(X_train_encoded.shape)

print("\nOriginal Testing Shape:")
print(X_test.shape)

print("\nEncoded Testing Shape:")
print(X_test_encoded.shape)

print("\nTotal Encoded Features:")
print(len(encoded_feature_names))

print("\nEncoded Feature Names:")
print(encoded_feature_names.tolist())
```

```python
# ============================================================
# STEP 93: TRAIN RANDOM FOREST MODEL
# ============================================================

from sklearn.ensemble import RandomForestClassifier

# Create Random Forest classifier
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)

# Train model
rf_model.fit(
    X_train_encoded,
    y_train
)

# Generate predictions
y_train_pred = rf_model.predict(X_train_encoded)
y_test_pred = rf_model.predict(X_test_encoded)

# Generate probability predictions
y_test_proba = rf_model.predict_proba(X_test_encoded)[:, 1]

print("=" * 70)
print("RANDOM FOREST MODEL")
print("=" * 70)

print("\nModel Training Completed Successfully.")

print("\nNumber of Trees:")
print(rf_model.n_estimators)

print("\nNumber of Training Records:")
print(X_train_encoded.shape[0])

print("\nNumber of Testing Records:")
print(X_test_encoded.shape[0])

print("\nNumber of Features:")
print(X_train_encoded.shape[1])

print("\nSample Predictions:")
print(y_test_pred[:20])

print("\nSample High-Risk Probabilities:")
print(y_test_proba[:20].round(3))
```

```python
# ============================================================
# STEP 94: EVALUATE RANDOM FOREST MODEL
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Training performance
train_accuracy = accuracy_score(
    y_train,
    y_train_pred
)

# Testing performance
test_accuracy = accuracy_score(
    y_test,
    y_test_pred
)

test_precision = precision_score(
    y_test,
    y_test_pred
)

test_recall = recall_score(
    y_test,
    y_test_pred
)

test_f1 = f1_score(
    y_test,
    y_test_pred
)

test_roc_auc = roc_auc_score(
    y_test,
    y_test_proba
)

# Confusion matrix
cm = confusion_matrix(
    y_test,
    y_test_pred
)

print("=" * 70)
print("RANDOM FOREST MODEL EVALUATION")
print("=" * 70)

print("\nTraining Accuracy:")
print(round(train_accuracy, 4))

print("\nTesting Accuracy:")
print(round(test_accuracy, 4))

print("\nTesting Precision:")
print(round(test_precision, 4))

print("\nTesting Recall:")
print(round(test_recall, 4))

print("\nTesting F1-Score:")
print(round(test_f1, 4))

print("\nTesting ROC-AUC:")
print(round(test_roc_auc, 4))

print("\nAccuracy Gap:")
print(round(train_accuracy - test_accuracy, 4))

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=[
            'Lower Risk',
            'High Risk'
        ]
    )
)
```

```python
# ============================================================
# STEP 95: LOGISTIC REGRESSION BASELINE MODEL
# ============================================================

from sklearn.linear_model import LogisticRegression

# Create Logistic Regression model
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42,
    class_weight='balanced'
)

# Train model
lr_model.fit(
    X_train_encoded,
    y_train
)

# Predictions
lr_train_pred = lr_model.predict(X_train_encoded)
lr_test_pred = lr_model.predict(X_test_encoded)

# Probability predictions
lr_test_proba = lr_model.predict_proba(X_test_encoded)[:, 1]

print("=" * 70)
print("LOGISTIC REGRESSION BASELINE")
print("=" * 70)

print("\nModel Training Completed Successfully.")

print("\nTraining Accuracy:")
print(round(
    accuracy_score(y_train, lr_train_pred), 4
))

print("\nTesting Accuracy:")
print(round(
    accuracy_score(y_test, lr_test_pred), 4
))

print("\nTesting Precision:")
print(round(
    precision_score(y_test, lr_test_pred), 4
))

print("\nTesting Recall:")
print(round(
    recall_score(y_test, lr_test_pred), 4
))

print("\nTesting F1-Score:")
print(round(
    f1_score(y_test, lr_test_pred), 4
))

print("\nTesting ROC-AUC:")
print(round(
    roc_auc_score(y_test, lr_test_proba), 4
))

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        lr_test_pred
    )
)
```

```python
# ============================================================
# STEP 101: DATA-DRIVEN RISK BANDS
# ============================================================

# Calculate percentile-based thresholds
low_threshold = (
    risk_predictions[
        'Predicted_Risk_Probability'
    ].quantile(0.33)
)

high_threshold = (
    risk_predictions[
        'Predicted_Risk_Probability'
    ].quantile(0.67)
)

# Create data-driven risk bands
risk_predictions['Risk_Band'] = pd.cut(
    risk_predictions['Predicted_Risk_Probability'],
    bins=[
        -np.inf,
        low_threshold,
        high_threshold,
        np.inf
    ],
    labels=[
        'Lower Risk',
        'Medium Risk',
        'Higher Risk'
    ]
)

print("=" * 70)
print("DATA-DRIVEN RISK BAND ANALYSIS")
print("=" * 70)

print("\nLower Risk Threshold:")
print(round(low_threshold, 4))

print("\nHigher Risk Threshold:")
print(round(high_threshold, 4))

print("\nRisk Band Distribution:")
print(
    risk_predictions[
        'Risk_Band'
    ].value_counts()
    .sort_index()
)

print("\nRisk Band Percentages:")
print(
    (
        risk_predictions['Risk_Band']
        .value_counts(normalize=True)
        .sort_index()
        * 100
    ).round(2)
)

print("\nRisk Band Probability Summary:")

risk_band_summary = (
    risk_predictions
    .groupby(
        'Risk_Band',
        observed=False
    )['Predicted_Risk_Probability']
    .agg(
        Count='count',
        Minimum='min',
        Average='mean',
        Maximum='max'
    )
    .round(4)
)

display(risk_band_summary)
```
